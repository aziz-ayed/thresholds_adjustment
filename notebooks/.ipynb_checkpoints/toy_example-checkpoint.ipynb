# Toy Example: Calibration & Threshold Analysis

This notebook illustrates the main methods used in the paper using a **synthetic multi-center dataset**:

- Generation of a toy dataset with centers, risk scores, and outcomes  
- Calibrated incidence at a decision threshold with bootstrap confidence intervals  
- Change in eligibility under an "oracle" threshold (target incidence) with bootstrap CIs  
- Out-of-fold (OOF) nested bootstrap calibration comparison across centers  
- Publication-style plots summarizing:
  - Absolute error to target incidence by center  
  - Observed incidence at the guideline threshold  
  - PPV, sensitivity, specificity by center  

The code relies on the lightweight package in `src/calib_paper`.

```python
import os
import sys
import warnings

warnings.filterwarnings("ignore")

# Ensure we can import from ../src if not installed as a package
ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
SRC = os.path.join(ROOT, "src")
if SRC not in sys.path:
    sys.path.append(SRC)

import numpy as np
import pandas as pd

from thresh_paper.data import make_toy_dataset
from thresh_paper.metrics import (
    compute_eligibility_change_with_ci,
    get_calibrated_value_with_ci,
    compute_confusion_metrics,
)
from thresh_paper.bootstrap import (
    out_of_fold_bootstrap_calibration_comparison_top_centers,
)
from thresh_paper.plotting import (
    plot_absolute_bias_summary,
    create_three_panel_figure,
)

import matplotlib.pyplot as plt

# Make a directory for figures
FIG_DIR = os.path.join(ROOT, "figs")
os.makedirs(FIG_DIR, exist_ok=True)

FIG_DIR
```

## 1. Generate a toy multi-center dataset

We simulate:

- 10 centers  
- 2000 patients per center  
- Center-level heterogeneity in log-odds  
- A risk score (`probs`) and binary outcome (`golds`)  

This is purely synthetic and is intended only to illustrate the methods.


```python
df = make_toy_dataset(
    n_centers=10,
    n_per_center=5000,
    random_state=0,
)

df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nEvent rate by center:")
display(
    df.groupby("center")["golds"].mean().rename("event_rate").round(4).to_frame()
)

global_incidence = df["golds"].mean()
print(f"\nGlobal event rate (target incidence) = {global_incidence:.4f}")
```

## 2. Eligibility change under an “oracle” threshold

We compare:

- A **guideline threshold** – here, for illustration, we take it equal to the *global event rate*
- An **oracle threshold** – the threshold derived from the calibration curve that achieves the target incidence

We then compute the **relative change** in the eligible population (proportion above threshold) with bootstrap confidence intervals.

```python
guideline_threshold = global_incidence / 2
target_incidence = global_incidence / 2

elig_stats = compute_eligibility_change_with_ci(
    golds=df["golds"],
    probs=df["probs"],
    guideline_threshold=guideline_threshold,
    target_incidence=target_incidence,
    n_bins=10,
    n_bootstrap=500,
    confidence_level=0.90,
    random_state=0,
)

elig_stats
```

## 3. Calibrated incidence at a threshold, with CIs

As an example, we examine the **subpopulation with age ≤ 65**, and estimate:

\\[
m(t) = \mathbb{E}[Y \mid \hat{p} \approx t]
\\]

at the decision threshold \\[ t = \text{guideline threshold} \\], using:

- A calibration curve  
- Bootstrap confidence intervals

This mimics the kind of local calibration checks often reported in the manuscript.


```python
younger = df[df["age"] <= 65]

cal_stats = get_calibrated_value_with_ci(
    golds=younger["golds"],
    probs=younger["probs"],
    threshold=guideline_threshold,
    n_bins=15,
    strategy="quantile",
    n_bootstrap=500,
    confidence_level=0.95,
    random_state=0,
)

cal_stats
```

## 4. Out-of-fold nested bootstrap calibration comparison across centers

We now:

1. Restrict to the **top 10 centers** (all centers in this toy dataset)
2. Run the **out-of-fold nested bootstrap** calibration comparison
3. Summarize, per center, the mean absolute error to the target incidence for multiple methods:
   - No adjustment / guideline threshold (`at_op`)
   - Threshold shifting (`intercept`)
   - Isotonic scaling (`iso`)
   - etc.

For the toy example we use a smaller number of outer bootstrap replicates to keep runtime reasonable.

```python
# The function expects center column name; rename for compatibility
df_boot = df.rename(columns={"center": "cen"})

# Import the plain get_calibrated_value_at_threshold used internally
from calib_paper.metrics import get_calibrated_value_at_threshold

summary_df = out_of_fold_bootstrap_calibration_comparison_top_centers(
    df=df_boot,
    cen_col="cen",
    score_col="probs",
    label_col="golds",
    target_incidence=target_incidence,
    get_calibrated_value_at_threshold=get_calibrated_value_at_threshold,
    top_k=10,
    nb_recal=500,    # dev-set size; can be tuned
    n_bins=20,
    n_outer=300,     # keep small for the example
    random_state=42,
    ci_width=0.90,
    use_direct_incidence=False,
)

summary_df = summary_df.rename(columns={"center": "center"})
summary_df.head()
```

## 5. Plot 1 – Absolute error to target incidence by center

This plot shows, for each center:

- Mean absolute deviation between observed incidence and target incidence  
- Bootstrap confidence intervals  

for several calibration strategies.


```python
fig1, ax1 = plot_absolute_bias_summary(
    summary_df,
    methods=["at_op", "iso", "intercept"],
    ci_width=0.95,
    show_pvalues=True,
    plot_type="abs_error",
    target_incidence=target_incidence,
    plot_title="Toy Data: Absolute Error to Target Incidence by Center",
)

fig_path1 = os.path.join(FIG_DIR, "toy_absolute_error_by_center.png")
fig1.savefig(fig_path1, dpi=300, bbox_inches="tight")
fig_path1
```

## 6. Plot 2 – Observed incidence at the guideline threshold (by center)

Here, for the **no-adjustment (at_op)** strategy, we show:

- Observed incidence at the guideline threshold  
- Bootstrap confidence intervals  
- Target incidence (horizontal dashed line)

```python
fig2, ax2 = plot_absolute_bias_summary(
    summary_df,
    methods=["at_op"],
    ci_width=0.95,
    show_pvalues=False,
    plot_type="threshold",
    target_incidence=target_incidence,
    plot_title="Toy Data: Observed Incidence at Guideline Threshold by Center",
)

fig_path2 = os.path.join(FIG_DIR, "toy_incidence_by_center.png")
fig2.savefig(fig_path2, dpi=300, bbox_inches="tight")
fig_path2
```

## 7. Plot 3 – PPV, Sensitivity, Specificity by center

Finally, we show standard performance metrics at the guideline threshold:

- Positive Predictive Value (PPV)  
- Sensitivity  
- Specificity  

for each center, in a three-panel figure.


```python
rows = []
for center, sub in df.groupby("center"):
    sens, spec, ppv = compute_confusion_metrics(
        golds=sub["golds"].values,
        probs=sub["probs"].values,
        threshold=guideline_threshold,
    )
    rows.append(
        {
            "center": center,
            "ppv": ppv,
            "sensitivity": sens,
            "specificity": spec,
        }
    )

metrics_df = pd.DataFrame(rows).sort_values("center")
metrics_df
```

```python
fig3, axes3 = create_three_panel_figure(
    df=metrics_df,
    suptitle="Toy Data: PPV, Sensitivity, Specificity at Guideline Threshold",
    plot_ci=False,
)

fig_path3 = os.path.join(FIG_DIR, "toy_three_panel_metrics.png")
fig3.savefig(fig_path3, dpi=300, bbox_inches="tight")
fig_path3
```

- How to quantify **eligibility change** when switching to an oracle threshold  
- How to run the **out-of-fold nested bootstrap** calibration comparison across centers  
- How to reproduce the main **publication-style plots** used in the paper
